**LSCA2318 · Tema 2.** Antes de que un modelo calcule vectores, tú los vas a construir a mano para entender qué significa «parecido = cerca».

**Qué haces:**
1. Eliges **4 a 6 dimensiones con nombre** que tengan sentido en tu variante (ej. `seguridad, pruebas, dinero, despliegue`).
2. Escribes **3 frases** de tu variante: F1 y F2 del mismo tema, F3 de otro tema.
3. Les das un valor de 0 a 1 a cada dimensión (justifícalo en una línea).
4. Normalizas y calculas el coseno F1–F2 y F1–F3. **Interpretas** el resultado.

Funciona en Colab o en VS Code. No necesita internet ni llave.

```python
import numpy as np

# Dimensiones temáticas para v2-revisor (Seguridad y Revisión de Código)
dimensiones = ["secretos", "sql_inyeccion", "privacidad_logs", "validacion_xss"]

# Definición de 3 vectores/frases
f1 = [0.90, 0.05, 0.05, 0.00]
f2 = [0.85, 0.10, 0.05, 0.00]
f3 = [0.05, 0.05, 0.10, 0.80]

# Diccionario con los textos y vectores
frases = {
    "F1": ("Cómo evitar claves de API expuestas en el código", f1),
    "F2": ("Proteger contraseñas y tokens en repositorios", f2),
    "F3": ("Sanitizar entradas de usuario para prevenir XSS", f3),
}

for k, (texto, v) in frases.items():
    assert len(v) == len(dimensiones), f"{k} debe tener {len(dimensiones)} valores"
    print(k, texto, dict(zip(dimensiones, v)))
```

**Justificación de tus valores** (doble clic para editar): explica en una línea por qué F1 tiene 0.9 en «seguridad», etc.

- F1: Concentra 0.90 en secretos porque trata directamente de proteger claves de API expuestas.
- F2: Tiene 0.85 en secretos y 0.10 en sql_inyeccion al enfocarse en tokens y contraseñas.
- F3: Concentra 0.80 en validacion_xss al tratar sobre sanitización de entradas para prevenir XSS.

```python
def normalizar(v):
    v = np.asarray(v, dtype=float)
    n = np.linalg.norm(v)
    return v / n if n > 0 else v      # el vector cero se queda en cero

def coseno(a, b):
    a, b = normalizar(a), normalizar(b)
    return float(a @ b)

V = {k: v for k, (_, v) in frases.items()}
for k in V:
    print(f"{k} normalizado:", np.round(normalizar(V[k]), 3), " norma =", round(np.linalg.norm(normalizar(V[k])), 3))

print(f"\ncos(F1, F2) = {coseno(V['F1'], V['F2']):.3f}   ← mismo tema")
print(f"cos(F1, F3) = {coseno(V['F1'], V['F3']):.3f}   ← temas distintos")
```

**Interpretación y Respuestas:**

1. **¿Salió F1–F2 mayor que F1–F3? ¿Por qué?**
   Sí, `cos(F1, F2) = 0.998` es significativamente mayor que `cos(F1, F3) = 0.072`. Esto sucede porque F1 y F2 comparten un alto peso en la dimensión de `secretos` (0.90 y 0.85 respectivamente), lo que orienta sus vectores casi en la misma dirección. En cambio, F3 enfoca su peso (0.80) en `validacion_xss`, generando un vector casi ortogonal.
2. **¿Qué dimensión pesó más en el resultado?**
   La dimensión de **`secretos`**, ya que fue el eje donde F1 y F2 concentraron la gran mayoría de su magnitud.
3. **¿Qué ganas y qué pierdes frente a dimensiones con nombre?**
   - **Ganas:** Explicabilidad directa e interpretabilidad humana (puedes saber exactamente por qué dos frases son parecidas leyendo las etiquetas).
   - **Pierdes:** Capacidad de representación semántica fina; los modelos reales usan espacios latentes de alta dimensión sin nombre que capturan matices complejos, sinónimos y contexto que no se pueden clasificar en 4 categorías rígidas.
